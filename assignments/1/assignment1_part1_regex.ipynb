```python
import os, re, urllib.request
import pandas as pd
import numpy as np
from datetime import datetime

def local_file(name):
    path = os.path.join("data", name)
    if not os.path.exists(path):
        os.makedirs("data", exist_ok=True)
        urllib.request.urlretrieve(f"https://raw.githubusercontent.com/ababacaryoro/nlp-course/main/assignments/1/data/{name}", path)
    return path

with open(local_file("dates.txt"), encoding="utf-8") as f:
    df = pd.Series([line for line in f])
print(f"{len(df)} notes")
df.head()
```

## Question 1 (1 point)

Return the list of every date written as `MM/DD/YY` or `MM/DD/YYYY` (for example `03/25/93`, `6/18/85`, `5/24/1990`) found in the notes.

```python
import re

def question_one():
    """List every MM/DD/YY or MM/DD/YYYY date string found in the notes."""
    pattern = r"\b\d{1,2}/\d{1,2}/(?:\d{2}|\d{4})\b"
    results = []

    for note in df:
        results.extend(re.findall(pattern, str(note)))

    return results

q1 = question_one()
print(f"{len(q1)} dates, first 10: {q1[:10]}")
```

## Question 2 (1 point)

Return the list of every date that starts with a month name followed by a day and a year: `Mar-20-2009`, `Mar 20, 2009`, `March 20, 2009`, `Mar. 20, 2009`, `Mar 20 2009`, `Mar 20th, 2009`.

```python
import re

def question_two():
    """List all 'Month day, year' dates found in the notes."""
    pattern = (
        r"\b(?:January|February|March|April|May|June|July|August|September|October|November|December"
        r"|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Sept|Oct|Nov|Dec)\.?"
        r"\s+\d{1,2},\s+\d{4}\b"
    )

    results = []
    for note in df:
        results.extend(re.findall(pattern, str(note), flags=re.IGNORECASE))
    return results

q2 = question_two()
print(f"{len(q2)} dates, first 10: {q2[:10]}")
```

## Question 5 (1 point)

`question_five(text)` cleans a text: remove digits, remove punctuation, lowercase, normalize whitespace.

```python
import string
import re

def question_five(text):
    """Cleaned text: no digit, no punctuation, lowercase, single spaces."""
    text = text.lower()
    text = re.sub(r"\d", "", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    return re.sub(r"\s+", " ", text).strip()

print(repr(question_five("Hello, World! 123 This is a TEST... with 456 numbers!!!")))
# Expected: 'hello world this is a test with numbers'
```

## Question 6 (2 points)

`question_six(text)` extracts the phone numbers written as `XXX-XXX-XXXX`, `(XXX) XXX-XXXX`, `XXX.XXX.XXXX` or `XXX XXX XXXX` and returns them normalized as `XXX-XXX-XXXX`. Sequences such as `12-34-5678` or `1234567890` are not valid.

```python
def question_six(text):
    """List of phone numbers normalized as XXX-XXX-XXXX."""
    pattern = r"(?<!\d)(?:\(?(\d{3})\)?[-.\s])(\d{3})[-.\s](\d{4})(?!\d)"
    matches = re.findall(pattern, text)
    return [f"{area}-{middle}-{last}" for area, middle, last in matches]

test_text = """
Call us at 123-456-7890 or (555) 123-4567.
You can also reach us at 888.555.1234 or 999 888 7777.
Invalid: 12-34-5678, 1234567890
"""
print(question_six(test_text))
# Expected: ['123-456-7890', '555-123-4567', '888-555-1234', '999-888-7777']
```

## Question 7 (3 points)

Extract the date of every note and sort the notes chronologically.

**Rules**
- `xx/xx/xx` is `mm/dd/yy`.
- A 2-digit year is in the 1900s (`1/5/89` is January 5th, 1989).
- When the day is missing (`9/2009`, `Feb 2009`), use the first day of the month.
- When the month is missing (`2010`), use January 1st.

`question_seven()` returns a `pandas.Series` of length 500 whose **values are the original indices** of the notes, in ascending chronological order.

Example: if the notes were `0: "1999"`, `1: "2010"`, `2: "1978"`, the function returns the Series `[2, 0, 1]`.

```python
def question_seven():
    """Return the note indices sorted by the earliest date found in each note."""
    # Regular expressions for the various date formats present in the dataset
    r1 = r'(\d{1,2})[/\-](\d{1,2})[/\-](\d{2,4})'
    r2 = r'((?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)[a-z]*[-.]?\s?\d{1,2}[a-z]*[-.,]?\s?\d{4})'
    r3 = r'(\d{1,2}\s(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)[a-z]*[-.,]?\s?\d{4})'
    r4 = r'((?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)[a-z]*[-.,]?\s?\d{4})'
    r5 = r'(\d{1,2}/\d{4})'
    r6 = r'(\d{4})'

    # Dictionary to map English month abbreviations/names to month numbers
    months = {
        'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6,
        'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12
    }

    parsed_dates = {}

    for idx, text in df.items():
        text = str(text).strip()
        matched_date = None

        # 1. MM/DD/YY or MM/DD/YYYY or MM-DD-YY or MM-DD-YYYY
        m1 = re.search(r1, text)
        if m1:
            m, d, y = m1.groups()
            month = int(m)
            day = int(d)
            year = int(y)
            if year < 100:
                year += 1900
            try:
                matched_date = datetime(year, month, day)
            except ValueError:
                pass

        # 2. Month Day, Year or Month-Day-Year or Month Day Year
        if not matched_date:
            m2 = re.search(r2, text, re.IGNORECASE)
            if m2:
                parts = re.split(r'[-.,\s]+', m2.group(1))
                if len(parts) >= 3:
                    m_str, d_str, y_str = parts[0][:3].lower(), parts[1], parts[2]
                    month = months.get(m_str, 1)
                    day = int(''.join(c for c in d_str if c.isdigit()))
                    year = int(y_str)
                    try:
                        matched_date = datetime(year, month, day)
                    except ValueError:
                        pass

        # 3. Day Month Year
        if not matched_date:
            m3 = re.search(r3, text, re.IGNORECASE)
            if m3:
                parts = re.split(r'[-.,\s]+', m3.group(1))
                if len(parts) >= 3:
                    d_str, m_str, y_str = parts[0], parts[1][:3].lower(), parts[2]
                    month = months.get(m_str, 1)
                    day = int(d_str)
                    year = int(y_str)
                    try:
                        matched_date = datetime(year, month, day)
                    except ValueError:
                        pass

        # 4. Month Year
        if not matched_date:
            m4 = re.search(r4, text, re.IGNORECASE)
            if m4:
                parts = re.split(r'[-.,\s]+', m4.group(1))
                if len(parts) >= 2:
                    m_str, y_str = parts[0][:3].lower(), parts[1]
                    month = months.get(m_str, 1)
                    day = 1
                    year = int(y_str)
                    try:
                        matched_date = datetime(year, month, day)
                    except ValueError:
                        pass

        # 5. MM/YYYY
        if not matched_date:
            m5 = re.search(r5, text)
            if m5:
                m_str, y_str = m5.group(1).split('/')
                month = int(m_str)
                day = 1
                year = int(y_str)
                try:
                    matched_date = datetime(year, month, day)
                except ValueError:
                    pass

        # 6. YYYY
        if not matched_date:
            m6 = re.search(r6, text)
            if m6:
                year = int(m6.group(1))
                month = 1
                day = 1
                try:
                    matched_date = datetime(year, month, day)
                except ValueError:
                    pass

        # Fallback in case absolutely nothing matched (should not happen with our structured rules)
        if not matched_date:
            matched_date = datetime(1900, 1, 1)

        parsed_dates[idx] = matched_date

    # Sort indices by their resolved dates
    sorted_indices = sorted(parsed_dates.keys(), key=lambda x: parsed_dates[x])
    return pd.Series(sorted_indices)

q7 = question_seven()
print(
    f"Length: {len(q7)}, "
    f"first 10 indices: {list(q7.head(10))}, "
    f"last 10: {list(q7.tail(10))}"
)
```

```python
import re
import string
import pandas as pd
from datetime import datetime

class TextDatePipeline:
    def __init__(self, df_series):
        self.df = df_series

    def extract_numerical_dates(self):
        """Question 1: MM/DD/YY or MM/DD/YYYY"""
        pattern = r"\b\d{1,2}/\d{1,2}/(?:\d{2}|\d{4})\b"
        results = []
        for note in self.df:
            results.extend(re.findall(pattern, str(note)))
        return results

    def extract_month_day_year(self):
        """Question 2: Month Day, Year"""
        pattern = (
            r"\b(?:January|February|March|April|May|June|July|August|September|October|November|December"
            r"|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Sept|Oct|Nov|Dec)\.?"
            r"\s+\d{1,2},\s+\d{4}\b"
        )
        results = []
        for note in self.df:
            results.extend(re.findall(pattern, str(note), flags=re.IGNORECASE))
        return results

    def extract_day_month_year(self):
        """Question 3: Day Month Year"""
        pattern = r"\b\d{1,2}\s+(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)[a-z]*[.,]?\s+\d{4}\b"
        results = []
        for note in self.df:
            results.extend(re.findall(pattern, str(note)))
        return results

    def extract_emails(self, text):
        """Question 4: Email address extractor"""
        pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
        return re.findall(pattern, text)

    def clean_text(self, text):
        """Question 5: Punctuation, digits, lowercase, whitespace cleaner"""
        text = text.lower()
        text = re.sub(r"\d", "", text)
        text = text.translate(str.maketrans("", "", string.punctuation))
        return re.sub(r"\s+", " ", text).strip()

    def extract_and_normalize_phones(self, text):
        """Question 6: Phone number normalization"""
        pattern = r"(?<!\d)(?:\(?(\d{3})\)?[-.\s])(\d{3})[-.\s](\d{4})(?!\d)"
        matches = re.findall(pattern, text)
        return [f"{area}-{middle}-{last}" for area, middle, last in matches]

    def get_chronological_indices(self):
        """Question 7: Chronologically sorted note indices"""
        patterns = [
            (re.compile(r"\b\d{1,2}/\d{1,2}/(?:\d{4}|\d{2})\b"), ("%m/%d/%Y", "%m/%d/%y")),
            (re.compile(
                r"\b(?:January|February|March|April|May|June|July|August|September|October|November|December"
                r"|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Sept|Oct|Nov|Dec)\.?"
                r"\s+\d{1,2},\s+\d{4}\b", re.IGNORECASE
            ), ("%B %d, %Y", "%b %d, %Y")),
            (re.compile(
                r"\b\d{1,2}\s+"
                r"(?:January|February|March|April|May|June|July|August|September|October|November|December"
                r"|Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Sept|Oct|Nov|Dec)\.?"
                r"\s+\d{4}\b", re.IGNORECASE
            ), ("%d %B %Y", "%d %b %Y")),
        ]

        dated_indices = []
        for index, note in self.df.items():
            if pd.isna(note):
                continue
            found_dates = []
            text = str(note)
            for pattern, formats in patterns:
                for match in pattern.finditer(text):
                    date_text = match.group(0).replace(".", "")
                    for date_format in formats:
                        try:
                            found_dates.append(datetime.strptime(date_text, date_format))
                            break
                        except ValueError:
                            continue
            if found_dates:
                dated_indices.append((min(found_dates), index))

        dated_indices.sort(key=lambda item: item[0])
        return pd.Series([index for _, index in dated_indices])

# Initialize pipeline with global df
pipeline = TextDatePipeline(df)
print("Pipeline successfully built!")
```
